```python
from ultralytics import YOLO
import random
from google.colab import files
from PIL import Image
import io
import matplotlib.pyplot as plt
```

```python
model = YOLO("/content/best.pt")
```

```python
class_mapping = ["Paper", "Rock", "Scizzors"]
```

```python
class Jogador:
    def __init__(self, nome, escolha=None):
        self.nome = nome
        self.escolha = escolha
        self.pontuacao = 0
```

```python
def define_vencedor(jogador1, jogador2):
    escolha1 = jogador1.escolha
    escolha2 = jogador2.escolha
    print(f'{jogador1.nome}:{class_mapping[escolha1]} X {jogador2.nome}:{class_mapping[escolha2]}')
    if escolha1 == escolha2:
        return "empate"
    elif (escolha2 - escolha1) % 3 == 1:
        return jogador1
    else:
        return jogador2
```

#Para jogar jokenpo basta clicar no botão na saída da célula e carregar uma foto com sua jogada (pedra, papel e tesoura), a máquina gerará aleatoriamente uma jogada contrária. Para encerrar basta carregar uma foto sem uma jogada.
##Atenção, para papel a mão é fechada

```python
usuario = Jogador("Usuário")
sistema = Jogador("Sistema")
while True:
  sistema.escolha = random.choice(range(3))
  uploaded = files.upload()

  filename = next(iter(uploaded))
  imagem = Image.open(io.BytesIO(uploaded[filename]))

  previsao = model.predict(imagem)

  if len(previsao[0].boxes.cls) !=1:
    print("Foi feito diferente de uma escolha")
    print("O jogo será encerrado")
    break

  usuario.escolha = int(previsao[0].boxes.cls[0])
  print(class_mapping[usuario.escolha])
  vencedor = define_vencedor(jogador1 = usuario, jogador2 = sistema)

  if isinstance(vencedor, str):
          print(vencedor)
  else:
      vencedor.pontuacao += 1
      print(f'vencedor: {vencedor.nome}')

  print(f'Usuário: {usuario.pontuacao} | Sistema: {sistema.pontuacao}')
```
