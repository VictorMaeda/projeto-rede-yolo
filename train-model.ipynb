```python
from roboflow import Roboflow
import os
from ultralytics import YOLO
import shutil
import os
```

```python
rf = Roboflow(api_key="1EqV7tInJZ9yluCCB5mj")
project = rf.workspace("roboflow-58fyf").project("rock-paper-scissors-sxsw")
version = project.version(9)
dataset = version.download("yolov8")
```

```python
PATH = dataset.location
class_mapping = {"papel":0, "pedra": 1, "tesoura": 2}
RUNS_PATH = "/content/runs/detect"
```

```python
model = YOLO("yolov8s.pt")

model.train(
    data=PATH+"/data.yaml",
    epochs=30,
    imgsz=640,
    batch=32,
    freeze=10
)
```

```python
predicts_dir = os.path.join(RUNS_PATH, "predicts")
os.makedirs(predicts_dir, exist_ok=True)
```

```python
metrics = model.val(data=PATH+"/data.yaml", split="test")
```

```python
# Executa a previsão e salva no local desejado
results = model.predict(
    source=PATH+ "/test/images",  
    save=True,              
    save_txt=False,         
    conf=0.25,              
    imgsz=640,              
    project=predicts_dir,     
    name="predicts",        
    exist_ok=True          
)
```
